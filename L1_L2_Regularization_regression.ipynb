```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("patient_health_data.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df["smoking_status"]= df["smoking_status"].apply(lambda x: 1 if x == "Yes" else 0)
```

```python
df
```

```python
X = df.drop(["health_risk_score"],axis=1)
y = df["health_risk_score"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.25,random_state=11)
```

```python
lm = LinearRegression()
lm.fit(X_train,y_train)
lm.score(X_test,y_test)
```

```python
alpha = [0.001,0.01, 0.1, 1.0,10,100,1000]
```

```python
for al in alpha:
    lasso = Lasso(alpha=al)
    lasso.fit(X_train,y_train)
    score = lasso.score(X_test,y_test)
    print(f'for alpha {al} score is {score}')
```

```python
for al in alpha:
    ridge = Ridge(alpha=1.0)
    ridge.fit(X_train,y_train)
    score = ridge.score(X_test,y_test)
    print(f'for alpha {al} score is {score}')
```
